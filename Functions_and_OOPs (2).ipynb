**Question 1:** What is a function in Python? Explain how functions are created and called. Also
explain parameters, arguments, default parameters, *args, and **kwargs with suitable
examples.
A **function** is a reusable block of code that performs a specific task. It is created using `def` and called using its name.

```python
def greet(name):
    print("Hello", name)

greet("Pranav")
```

* **Parameter:** Variable defined in a function.
* **Argument:** Actual value passed to a function.
* **Default parameter:** Parameter with a default value.

```python
def greet(name="Guest"):
    print(name)
```

* **`*args`:** Accepts multiple positional arguments as a tuple.

```python
def add(*args):
    print(sum(args))
```

* **`**kwargs`:** Accepts multiple keyword arguments as a dictionary.

```python
def info(**kwargs):
    print(kwargs)
```

Functions make code **reusable and organized**.

**Question 2:** Create a Python program with a Person class containing the person's name and a
method to display it. Then create a Student class that inherits from Person and adds the
student's course. Create an object of Student and display both details.


```python
class Person:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print("Name:", self.name)


class Student(Person):
    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

    def display_details(self):
        self.display_name()
        print("Course:", self.course)


student = Student("Pranav", "Data Science")
student.display_details()
```

**Output:**

```text
Name: Pranav
Course: Data Science
```

**Question 3:**  What is variable scope in Python? Explain local and global variables. Why should
excessive use of the global keyword be avoided? Also explain nested functions and closures.

**Variable scope** defines where a variable can be accessed in a Python program.

* **Local Variable:** Created inside a function and accessible only within that function.
* **Global Variable:** Created outside a function and accessible throughout the program.
* **`global` keyword:** Allows a function to modify a global variable. Excessive use should be avoided because it can make code difficult to understand, debug, and maintain.
* **Nested Function:** A function defined inside another function.
* **Closure:** A nested function that remembers and accesses variables from its outer function even after the outer function has finished.


**Question 4:**  What is a lambda function? Explain how map(), filter(), and reduce() work in
Python with suitable examples.

A **lambda function** is a small anonymous function created using the `lambda` keyword.

```python
square = lambda x: x * x
print(square(5))  # 25
```

* **`map()`** – Applies a function to every item.

```python
print(list(map(lambda x: x * 2, [1, 2, 3])))
# [2, 4, 6]
```

* **`filter()`** – Selects items based on a condition.

```python
print(list(filter(lambda x: x % 2 == 0, [1, 2, 3, 4])))
# [2, 4]
```

* **`reduce()`** – Combines all items into one result.

```python
from functools import reduce
print(reduce(lambda x, y: x + y, [1, 2, 3, 4]))
# 10
```

**Question 6:** Create a custom iterator class called EvenNumbers that generates even numbers
from 2 to 10.
Use __iter__(), __next__(), and StopIteration

```python
class EvenNumbers:
    def __init__(self):
        self.num = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.num > 10:
            raise StopIteration
        value = self.num
        self.num += 2
        return value

for n in EvenNumbers():
    print(n)
```

**Output:**

```text
2
4
6
8
10
```

**Question 8:** What is polymorphism? Explain it in your own words with a Python example where
two different classes have the same method name but perform different tasks.

**Polymorphism** means "many forms." In Python, different classes can have the **same method name** but perform different tasks.

```python id="2s7z6m"
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

Dog().sound()
Cat().sound()
```

**Output:**

```text id="2s8z1q"
Dog barks
Cat meows
```

Here, both classes use the same `sound()` method, but they perform different actions.

**Question 9:** Consider the following list:
numbers = [1, 2, 3, 4, 5, 6]
Write a Python program to:
1. Use map() to calculate the square of every number.
2. Use filter() to select only even numbers.
3. Use reduce() to calculate the sum of all numbers.
4. Display all three results.
```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

squares = list(map(lambda x: x * x, numbers))
even = list(filter(lambda x: x % 2 == 0, numbers))
total = reduce(lambda x, y: x + y, numbers)

print("Squares:", squares)
print("Even numbers:", even)
print("Sum:", total)
```

**Output:**

```text
Squares: [1, 4, 9, 16, 25, 36]
Even numbers: [2, 4, 6]
Sum: 21
```


**Question 10:** You are developing a payment system for an online shopping application. The
system should support UPI, credit card, and wallet payments. Each payment type should have
its own way of processing the payment, but the main program should be able to call the same
pay() method for all of them. Design a simple Python solution using inheritance and
polymorphism, and explain how it works.

```python
class Payment:
    def pay(self, amount):
        pass

class UPI(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI")

class CreditCard(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Credit Card")

class Wallet(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Wallet")

payments = [UPI(), CreditCard(), Wallet()]

for payment in payments:
    payment.pay(1000)
```

**Output:**

```text
Paid 1000 using UPI
Paid 1000 using Credit Card
Paid 1000 using Wallet
``
